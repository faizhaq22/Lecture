# Dictionary

Dictionary

- Tidak Memiliki Urutan
- Pengambilan item dengan cara menggunakan 'key'
- format -> dict1={'key1:item,'key2:item,'key':item}

List
- memiliki urutan
- pengambilan item dengan memanggil index-nya (urutan item)
- format -> =[item1,item2,item3]
