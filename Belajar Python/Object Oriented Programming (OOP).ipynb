```python
class kulkas:
    
    penjual = "Pak Budi"
    
    def __init__(self, merek, harga):
        self.merek = merek
        self.harga = harga
    
    def keterangan(self):
        print('Kulkas merek {} harganya sebesar {}, dan nama penjualnya {}'.format(self.merek,self.harga,self.penjual))
```

```python
barang3= kulkas('LG',200)
```

```python
barang3.penjual
```

```python
barang3.keterangan()
```

```python
class lingkaran:
    
    #koefisien phi 
    phi = 3.14
    
    def __init__ (self,radius):
        self.radius = radius
        
    def keliling(self):
        return 2*self.phi*self.radius
    
    def luas (self):
        return 2 *self.phi*self.radius*self.radius
```

```python
l1 = lingkaran(20)
```

```python
l1.keliling()
```
