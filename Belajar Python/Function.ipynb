**Args dan **Kwargs  ** 

args (argumen) yang digunakan variabel berapa saja (*)
kwargs ; keyword argumen (**)

```python
def jumlah (*args):
    return sum((args))
```

```python
jumlah (2,3,4,2,3,4,4)
```

```python
def cek_args(*args):
    print(args)
```

```python
cek_args(2,3,4,1)
```

```python
def cek_Kwargs(**Kwargs):
    print(Kwargs)
```

```python
cek_Kwargs(anak=10)
```

```python
def cek_Kwargs(**Kwargs):
    if 'Budi' in Kwargs :
        print('Berat budi tergolong kategori {}'.format((Kwargs['Budi'])))
    else: 
        print('budi tidak termausk')
```

```python
cek_Kwargs(Budi='Gemuk', Joko='Kurus')
```

Menggabungkan *args dan **kwargs

```python
def fungsiku(*args,**kwargs):
    print("saya ingin membeli {} {} secara tunai".format(args[0],kwargs['makanan']))
```

```python
fungsiku(23,43,2,21, buah ='apel', makanan='soto')
```
