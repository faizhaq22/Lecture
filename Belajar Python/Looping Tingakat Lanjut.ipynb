**Enumerate**

```python
indeks = 0 
for item in ('abcdefghij'):
    print('karakter {} berada di indeks ke - {}'. format(item,indeks))
    indeks +=1
```

```python
kata = 'abcdefghij'
for item in enumerate (kata):
    print (item)
```

```python
for a,b in enumerate (kata):
    print(a)
    print(b)
```

**ZIP**
digunakan untuk menggabungkan 1/lebih list

```python
list1=[1,2,3,4,5]
list2=['a','b','c','d','e']
```

```python
for item in zip(list1,list2):
    print(item)
```

List Comprehensif (Mempersingkat Kode)

```python
isi3=[item for item in range(1,11)]
isi3
```

```python
isi4 = [new**2 for new in range(1,11)]
```

```python
isi4
```

```python
isi5=[item for item in range(1,11) if item%2==0]
```

```python
isi5
```

```python
isi6=[item if item%2==0 else'angka ini bilangan genap' for item in range(1,11)]
```

```python
isi6
```

```python
isi7 = []
for item in range(1,11):
    if item%2==0:
        isi7.append(item)
    else:
        isi7.append('angka ini bilangan genap')
```

```python
isi7
```
