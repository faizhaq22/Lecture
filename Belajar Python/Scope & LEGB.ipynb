**Urutan Eksekusi Pyhton**
- L(local) = Variabel ini bersifat loka, yang biasanya berada di dalam fungsi (def/lambda) tertentu).
- E (Enclosing function local) = variabel ini berdada di dalam fungsi (def/lambda) mulai dari lingkung terdalam hingga keluar
- G ( Global) = variabel ini didefinisikan di luar, biasanya di tahap awal
- B (bulit in ) variabel bawaan python, seperti list, str dll

```python
print(jumlah())
```

```python
angka = 99

def tulis (angka):
      
    #mendefinisikan ulang sebuah variabel global di fungsi
    angka = "Ganti Angka"
    print(f'angkanya sekarang saya rubah menjadi {angka}')
    return angka
```

```python
angka = tulis(angka)
```

```python
print(angka)
```
